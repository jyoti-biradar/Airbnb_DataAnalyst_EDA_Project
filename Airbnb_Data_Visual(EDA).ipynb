## There are three important parts of Exploratory Data Analysis (EDA).
Three types of EDA

1. Univariate Analysis
One variable
Analyzes one column at a time to understand its distribution, frequency, or summary statistics.
Airbnb example: What is the distribution of property prices?

2. Bivariate Analysis
Two variables
Examines the relationship or comparison between two columns.
Airbnb example: How does price vary by room type?

3. Multivariate Analysis
Three or more variables
Examines patterns and relationships among three or more columns at the same time.
Airbnb example: How are price, room type, and neighbourhood related?

```python
''' 
## Charts:- 
Common types of charts in Data Visualization

Bar chart -------> 	Comparing categories
Line chart ------------> Showing changes over time
Histogram --------->	Understanding numerical distributions
Box plot ----------->	Comparing distributions and spotting potential outliers
Scatter plot -------->	Exploring relationships between two numerical variables
Pie chart --------------> Showing proportions of a small number of categories
Heatmap	--------------> Showing correlations or patterns in a matrix
'''
```

## 1) Univariate Analysis — one variable

##  A) Distribution of Price Column:- 

```python
plt.figure(figsize=(6, 4))
sns.histplot(data=df, x="Price($)", kde = True ,bins=30,color= "Blue")
plt.title("Univariate Analysis: Price Distribution")
plt.xlabel("Price")
plt.ylabel("Number of Listings")
plt.tight_layout()
plt.show()
```

## B) Distribution of Room Type:- 

```python
plt.figure(figsize=(6, 4))
sns.countplot(
    data=df,
    x="Room_Type",
    order=df["Room_Type"].value_counts().index
)
plt.title("Univariate Analysis: Room Type")
plt.xlabel("Room Type")
plt.ylabel("Number of Listings")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## C) Distribution of Neighbourhood

```python
plt.figure(figsize=(6, 4))
sns.countplot(
    data=df,
    x="Neighbourhood_Group",
    order=df["Neighbourhood_Group"].value_counts().index
)
plt.title("Univariate Analysis: Room Type")
plt.xlabel("Neighbourhood_Group")
plt.ylabel("Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## 2) Bivariate Analysis — two variables

## A) Price vs. room type
- Visualize the relationship between Price and Room type

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x="Room_Type", y="Price($)",hue ="Room_Type")
plt.title("Bivariate Analysis: Price vs Room Type")
plt.xlabel("Room Type")
plt.ylabel("Price (dataset units)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x="Room_Type", y="Price($)",hue = "Room_Type")
plt.title("Bivariate Analysis: Price vs Room Type")
plt.xlabel("Room Type")
plt.ylabel("Price (dataset units)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## B) Price vs Neighbourhood_Group

```python
plt.figure(figsize=(7,4))
sns.barplot(data=df, x='Neighbourhood_Group', y='Price($)',
            hue='Room_Type')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0)
plt.title("Bivariate Analysis: Price vs Neighbourhood_Group")
plt.xlabel("Neighbourhood_Group")
plt.ylabel("Price($)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## C) Room_Type vs Number_of_Reviews

```python
plt.figure(figsize=(6, 4))
plt.title("Bivariate Analysis: Number_of_Reviews vs Room_Type")
sns.scatterplot(data=df, x='Number_of_Reviews', 
                y='Room_Type', hue='Neighbourhood_Group')
plt.show()
```

## D) Price($)  vs Number_of_Reviews

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Number_of_Reviews",
    y="Price($)",
    hue= "Neighbourhood_Group"
)

plt.title("Bivariate Analysis: Price vs Number of Reviews")
plt.xlabel("Number of Reviews")
plt.ylabel("Price (dataset units)")
plt.tight_layout()
plt.show()
```

## 3) Multivariate Analysis — three or more variables

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Neighbourhood_Group",
    y="Price($)",
    hue="Room_Type"
)

plt.title(
    "Multivariate Analysis: Price by Neighbourhood and Room Type"
)
plt.xlabel("Neighbourhood Group")
plt.ylabel("Price (dataset units)")
plt.xticks(rotation=30)
plt.legend(title="Room Type", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```
